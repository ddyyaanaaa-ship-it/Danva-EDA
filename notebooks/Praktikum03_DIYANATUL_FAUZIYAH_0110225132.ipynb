## Bagian E — Menerapkan tiga tindakan

```python
# TODO E1 — HAPUS
# Buat DataFrame baru bernama 'aman' yang sudah membuang semua kolom bertindakan 'hapus'.
# Cetak berapa kolom yang dibuang dan berapa yang tersisa.

kolom_hapus = [k for k, (_, t) in PENANDAAN.items() if t == 'hapus']
aman = df.drop(columns=kolom_hapus).copy()

print('Kolom dibuang :', len(kolom_hapus), kolom_hapus)
print('Kolom tersisa :', aman.shape[1])
```

```python
# TODO E2 — HASH
# Tulis fungsi samarkan(nilai, garam) yang mengembalikan 12 karakter pertama
# dari sha256(garam + nilai). Kembalikan np.nan bila nilainya kosong.
# Terapkan pada kolom yang Anda beri tindakan 'hash'.
# Expected: nilai asli dan hasil hash berbeda, dan nilai asli yang sama menghasilkan hash yang sama

# Garam dibaca dari variabel lingkungan supaya tidak tertulis di notebook (lihat daftar periksa G5).
# Nilai bawaan di bawah hanya untuk latihan kelas.
GARAM = os.environ.get('GARAM_TIM', 'ganti-dengan-kalimat-rahasia-tim')

def samarkan(nilai, garam):
    if pd.isna(nilai):
        return np.nan
    return hashlib.sha256((garam + str(nilai)).encode('utf-8')).hexdigest()[:12]

kolom_hash = [k for k, (_, t) in PENANDAAN.items() if t == 'hash']
for c in kolom_hash:
    aman[c] = df[c].apply(lambda x: samarkan(x, GARAM))

banding = pd.DataFrame({'asli': df['id_peserta'], 'hash': aman['id_peserta']})
print(banding.head(3))
print('Nilai asli != hash        :', (banding['asli'] != banding['hash']).all())
print('Jumlah ID unik asli / hash:', banding['asli'].nunique(), '/', banding['hash'].nunique())
```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca
# a. Hash dua kali nilai yang sama, tunjukkan hasilnya identik
# b. Hash dua nilai berbeda, tunjukkan hasilnya berbeda

# a. nilai yang sama -> hash identik
h1 = samarkan('PLT-2025-1309', GARAM)
h2 = samarkan('PLT-2025-1309', GARAM)
print(h1, h2, '-> identik:', h1 == h2)

# b. nilai berbeda -> hash berbeda
h3 = samarkan('PLT-2025-1310', GARAM)
print(h1, h3, '-> berbeda:', h1 != h3)
```

**Jawaban Anda:** kalau `GARAM` ikut ter-*commit* ke GitHub, apa yang masih terlindungi dan apa yang tidak?

**Yang masih terlindungi:** nilai asli tidak tersimpan di berkas mana pun, dan SHA-256 bersifat satu arah, jadi hash tidak bisa "dibalik" lewat rumus. Orang yang hanya memegang data hasil hash tanpa tahu garamnya tetap tidak bisa menebak ID aslinya.

**Yang tidak terlindungi:** garam adalah satu-satunya rahasia dalam skema ini. Begitu garam bocor, siapa pun dapat menebak semua nilai asli dengan cara mencoba kandidat satu per satu (serangan kamus/*brute force*), dan ID berpola seperti `PLT-2025-xxxx` hanya punya ribuan kemungkinan sehingga habis dicoba dalam hitungan detik. Lebih buruk lagi, garam yang sudah masuk riwayat Git tidak bisa ditarik kembali. Garam harus diganti dan seluruh hash dibuat ulang, tetapi hash lama yang sudah terunggah tetap tertinggal di riwayat.

```python
# TODO E4 — GENERALISASI
# a. Hitung usia peserta pada tanggal acuan 1 September 2025
# b. Buat kolom 'kelompok_umur' dengan batas: <25, 25-34, 35-44, 45-54, 55+
# c. Buang kolom tanggal_lahir dan kelurahan dari 'aman'
# d. Rapikan kolom kecamatan yang sebagian tertulis KAPITAL
# Expected: kelompok terbanyak adalah 35-44; kecamatan punya tepat 10 nilai unik

ACUAN = pd.Timestamp('2025-09-01')

# a. usia pada tanggal acuan
tl = aman['tanggal_lahir']
sudah_ulang_tahun = (tl.dt.month < ACUAN.month) | ((tl.dt.month == ACUAN.month) & (tl.dt.day <= ACUAN.day))
aman['usia'] = ACUAN.year - tl.dt.year - (~sudah_ulang_tahun).astype(int)

# b. kelompok umur
aman['kelompok_umur'] = pd.cut(aman['usia'], bins=[-np.inf, 24, 34, 44, 54, np.inf],
                               labels=['<25', '25-34', '35-44', '45-54', '55+'])

# c. buang tanggal_lahir dan kelurahan; usia eksak juga dibuang karena sama bocornya dengan tanggal lahir
aman = aman.drop(columns=['tanggal_lahir', 'kelurahan', 'usia'])

# d. rapikan kecamatan yang sebagian tertulis KAPITAL
aman['kecamatan'] = aman['kecamatan'].str.title()

print(aman['kelompok_umur'].value_counts(sort=False))
print('Kelompok terbanyak :', aman['kelompok_umur'].value_counts().idxmax())
print('Nilai unik kecamatan:', aman['kecamatan'].nunique())
```

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F1 — tulis fungsi cek_k(data, kolom)
# Kembalikan Series berisi: k (grup terkecil), grup berisi 1 orang, total grup

def cek_k(data, kolom):
    ukuran = data.groupby(kolom, observed=True).size()
    return pd.Series({'k': int(ukuran.min()),
                      'grup_1_orang': int((ukuran == 1).sum()),
                      'total_grup': int(len(ukuran))})
```

```python
# TODO F2 — ukur risiko SEBELUM penanganan
# Gunakan data asli df dengan kolom:
#   ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
# Expected: k = 1 dan SELURUH 620 baris berada di grup berisi satu orang

QI_SEBELUM = ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
print(cek_k(df, QI_SEBELUM))
```

**Jawaban Anda:** keempat kolom itu tidak memuat nama maupun nomor telepon. Jelaskan dalam dua kalimat mengapa hasilnya tetap berbahaya.

Kombinasi kelurahan, jenis kelamin, pendidikan, dan tanggal lahir lengkap bersifat unik untuk setiap dari 620 peserta (k = 1), sehingga siapa pun yang mengetahui keempat hal itu tentang seseorang (misalnya dari media sosial atau data publik lain) dapat menunjuk tepat satu baris tanpa perlu nama. Dengan menautkan ke sumber lain (*linkage*), baris itu langsung membuka data spesifik orang tersebut seperti penghasilan pribadi dan status disabilitas, jadi menghapus nama saja tidak menghilangkan risiko re-identifikasi.

```python
# TODO F4 — TANTANGAN: capai k >= 5
# Generalisasikan lebih jauh sampai k >= 5, dengan MEMPERTAHANKAN sebanyak mungkin informasi.
# Anda boleh: memperlebar kelompok umur, menggabungkan kecamatan kecil menjadi 'Lainnya',
# atau mengeluarkan satu kolom dari daftar quasi-identifier.
# Catat setiap pilihan dan dampaknya terhadap k.
# Petunjuk: ada lebih dari satu jawaban benar. Yang dinilai adalah alasannya.

# Strategi: mulai dari kombinasi F3, lalu longgarkan satu langkah demi satu langkah dan catat k-nya.
d = aman.copy()

# Langkah 1: perlebar kelompok umur 5 -> 3 kelompok
d['umur_3'] = d['kelompok_umur'].map({'<25': '<35', '25-34': '<35', '35-44': '35-44', '45-54': '45+', '55+': '45+'})

# Langkah 2: gabungkan kecamatan kecil menjadi 'Lainnya' (pertahankan 4 kecamatan terbesar)
besar = aman['kecamatan'].value_counts().index[:4]
d['kecamatan_gab'] = np.where(d['kecamatan'].isin(besar), d['kecamatan'], 'Lainnya')

# Langkah 3 (alternatif): pendidikan 6 jenjang -> 3 kelompok
d['pendidikan_3'] = d['pendidikan_terakhir'].map({'SD': 'SMP ke bawah', 'SMP': 'SMP ke bawah', 'SMA/SMK': 'SMA/SMK',
                                                  'D3': 'D3 ke atas', 'S1': 'D3 ke atas', 'S2': 'D3 ke atas'})

langkah = {
    '0. Awal (F3)':                                    ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur'],
    '1. Umur jadi 3 kelompok':                         ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'umur_3'],
    '2. + kecamatan digabung':                         ['kecamatan_gab', 'jenis_kelamin', 'pendidikan_terakhir', 'umur_3'],
    '3. + pendidikan jadi 3 kelompok':                 ['kecamatan_gab', 'jenis_kelamin', 'pendidikan_3', 'umur_3'],
    '4. Keluarkan pendidikan (PILIHAN)':               ['kecamatan_gab', 'jenis_kelamin', 'umur_3'],
    'Alternatif: keluarkan kecamatan':                 ['jenis_kelamin', 'pendidikan_3', 'umur_3'],
    'Dicoba: tanpa kecamatan, pend. 6 & umur 5 (gagal)': ['jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur'],
}
jejak = pd.DataFrame({nama: cek_k(d, kol) for nama, kol in langkah.items()}).T
print(jejak)

# Terapkan pilihan akhir pada data yang akan dirilis.
# Penting: pendidikan_terakhir HARUS ikut dibuang dari data rilis, bukan sekadar dari daftar uji,
# karena kolom yang tetap ada di data tetap bisa dipakai untuk menautkan orang.
QI_PILIHAN = ['kecamatan_gab', 'jenis_kelamin', 'umur_3']
aman_k5 = (d.drop(columns=['kecamatan', 'kelompok_umur', 'pendidikan_terakhir', 'pendidikan_3'])
             .rename(columns={'kecamatan_gab': 'kecamatan', 'umur_3': 'kelompok_umur'}))
print()
print('Data rilis:', aman_k5.shape, '| k pada data rilis:')
print(cek_k(aman_k5, ['kecamatan', 'jenis_kelamin', 'kelompok_umur']))
```

**Jawaban Anda:** tulis kombinasi yang Anda pilih, nilai k yang dicapai, dan informasi apa yang Anda korbankan untuk mencapainya.

**Kombinasi pilihan:** `kecamatan` (empat terbesar: Pancoran Mas, Sukmajaya, Cimanggis, Beji; sisanya digabung menjadi *Lainnya*) + `jenis_kelamin` + `kelompok_umur` tiga tingkat (<35, 35-44, 45+). Hasilnya **k = 7** pada 30 grup.

**Jejak keputusan (lihat tabel `jejak`):**
1. Memperlebar umur dari 5 ke 3 kelompok dan menggabungkan kecamatan kecil menjadi *Lainnya* menurunkan banyak grup tunggal, tetapi k masih 1.
2. Menyederhanakan pendidikan dari 6 jenjang menjadi 3 kelompok juga belum cukup: k tetap 1. Pendidikan adalah kolom dengan banyak kategori kecil (SD, S2), jadi kombinasinya terus memecah data menjadi grup mungil.
3. k ≥ 5 baru tercapai setelah **pendidikan dikeluarkan** dari daftar quasi-identifier dan dari data rilis.

**Informasi yang dikorbankan:** (a) pendidikan terakhir hilang sama sekali dari data rilis, jadi analisis "skor menurut pendidikan" tidak bisa dilakukan dari berkas rilis; (b) umur hanya tersisa tiga kelompok, sehingga perbedaan antara usia 25 dan 34 tahun tidak terlihat lagi; (c) enam kecamatan kecil melebur menjadi *Lainnya*, sehingga perbandingan antarkecamatan kecil tidak mungkin. Variabel analisis utama (kategori usaha, omzet, skor, kehadiran) tidak terganggu.

**Mengapa saya memilih ini, bukan alternatif:** membuang kecamatan (alternatif di tabel) juga bisa mencapai k ≥ 5 dan mempertahankan pendidikan, tetapi pertanyaan analisis kelas ini (bagian A) lebih dekat ke sebaran per kategori usaha dan wilayah binaan, sehingga kecamatan dinilai lebih berharga daripada pendidikan. Pilihan lain yang sama sahnya: pertahankan pendidikan (3 kelompok) dan umur (3 kelompok), lalu buang kecamatan (k = 11). Mempertahankan pendidikan 6 jenjang dan umur 5 kelompok tanpa kecamatan ternyata gagal (k = 1). Yang menentukan adalah pertanyaan analisis tim.

**Catatan sisa risiko:** uji k di atas hanya menilai tiga kolom quasi-identifier. Kolom numerik lain seperti `penghasilan_pribadi_bulanan` dan `jumlah_anak` tetap bisa membedakan orang bila dirilis per baris, sehingga sebaiknya dilaporkan sebagai ringkasan, bukan nilai per orang.

```python
import pandas as pd, numpy as np, hashlib, os, glob, getpass

# --- cari file otomatis ---
cands = glob.glob('*.csv')
print('CSV di folder:', cands)
FILE = next((f for f in cands if 'supermarket' in f.lower()), None)
if FILE is None:
    from google.colab import files
    up = files.upload()
    FILE = list(up.keys())[0]
print('Memakai:', FILE)

raw = pd.read_csv(FILE)
raw.columns = raw.columns.str.strip()
print(raw.columns.tolist())
df = raw.copy()
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
print('Tanggal gagal terbaca:', df['Date'].isna().sum())
df['Jam'] = df['Time'].astype(str).str.split(':').str[0].astype(int)

# --- G3: KAMUS DATA ---
manual = {
 'Invoice ID': ('Nomor faktur transaksi, unik per baris.', 'nominal', 'pengenal langsung (pseudonim)', 'hash'),
 'Branch': ('Kode cabang (A, B, C); satu-satu dengan City.', 'nominal', 'quasi-identifier', 'hapus'),
 'City': ('Kota lokasi cabang supermarket.', 'nominal', 'quasi-identifier', 'pertahankan'),
 'Customer type': ('Status pelanggan: Member atau Normal.', 'nominal', 'quasi-identifier', 'pertahankan'),
 'Gender': ('Jenis kelamin pelanggan.', 'nominal', 'quasi-identifier', 'pertahankan'),
 'Product line': ('Kategori produk yang dibeli.', 'nominal', 'bukan data pribadi', 'pertahankan'),
 'Unit price': ('Harga satu unit barang (USD).', 'rasio', 'bukan data pribadi', 'pertahankan'),
 'Quantity': ('Jumlah unit barang yang dibeli.', 'rasio', 'bukan data pribadi', 'pertahankan'),
 'Tax 5%': ('Pajak 5% dari cogs (USD); nilainya sama dengan gross income.', 'rasio', 'bukan data pribadi', 'pertahankan'),
 'Total': ('Total tagihan termasuk pajak (USD).', 'rasio', 'bukan data pribadi', 'pertahankan'),
 'Date': ('Tanggal transaksi (Jan-Mar 2019).', 'interval', 'quasi-identifier', 'generalisasi'),
 'Time': ('Jam transaksi (10:00-20:59).', 'interval', 'quasi-identifier', 'generalisasi'),
 'Payment': ('Metode pembayaran: Cash, Credit card, Ewallet.', 'nominal', 'bukan data pribadi', 'pertahankan'),
 'cogs': ('Harga pokok penjualan = Unit price x Quantity (USD).', 'rasio', 'bukan data pribadi', 'pertahankan'),
 'gross margin percentage': ('Persentase margin kotor; konstan di seluruh baris.', 'rasio', 'bukan data pribadi', 'pertahankan'),
 'gross income': ('Pendapatan kotor per transaksi (USD); sama dengan Tax 5%.', 'rasio', 'bukan data pribadi', 'pertahankan'),
 'Rating': ('Penilaian kepuasan pelanggan, skala 1-10.', 'ordinal', 'bukan data pribadi', 'pertahankan'),
}
kosong = ('(isi manual)', '', '', '')
print('Kolom tak terdaftar di kamus:', [c for c in raw.columns if c not in manual])

kamus = pd.DataFrame({
    'nama_kolom'  : raw.columns,
    'tipe_data'   : raw.dtypes.astype(str).values,
    'nilai_unik'  : raw.nunique().values,
    'nilai_kosong': raw.isna().sum().values,
    'contoh_nilai': [raw[c].iloc[0] for c in raw.columns],
})
for i, k in enumerate(['deskripsi','skala_pengukuran','kategori_data_pribadi','tindakan_penanganan']):
    kamus[k] = [manual.get(c, kosong)[i] for c in kamus['nama_kolom']]
kamus.to_csv('data_dictionary.csv', index=False)
print(kamus[['nama_kolom','kategori_data_pribadi','tindakan_penanganan']].to_string(index=False))

# --- G4: PENANGANAN ---
def cek_k(data, kolom):
    g = data.groupby(kolom, observed=True, dropna=False).size()
    return pd.Series({'k (grup terkecil)': g.min(),
                      'grup berisi 1 orang': (g == 1).sum(),
                      'total grup': len(g)})

print('\nk SEBELUM:')
print(cek_k(df, ['City','Gender','Customer type','Date','Time']))

GARAM = os.environ.get('GARAM_TIM') or getpass.getpass('Garam hash (jangan ditulis di notebook): ')
def samarkan(nilai, garam):
    if pd.isna(nilai): return np.nan
    return hashlib.sha256((garam + str(nilai)).encode('utf-8')).hexdigest()[:12]

aman = df.copy()
aman['Invoice ID'] = aman['Invoice ID'].apply(lambda v: samarkan(v, GARAM))
aman['Bulan'] = aman['Date'].dt.month
aman['Blok_waktu'] = pd.cut(aman['Jam'], bins=[0,12,15,24], labels=['<=12','13-15','16+'])
aman = aman.drop(columns=[c for c in ['Date','Time','Jam','Branch'] if c in aman.columns])

QI = ['City','Gender','Customer type','Blok_waktu','Bulan']
print('\nk SESUDAH:')
print(cek_k(aman, QI))

tab = pd.crosstab(aman['City'], aman['Product line'])
print('\nSel terisi <5:', ((tab > 0) & (tab < 5)).sum().sum())

os.makedirs('data/processed', exist_ok=True)
aman.to_csv('data/processed/supermarket_aman.csv', index=False)
print('\nTersimpan di data/processed/')
```
