## Bagian B — Seleksi dan filter (Latihan 2, berpasangan)
**Tujuan:** mengambil bagian data yang relevan dengan pertanyaan analisis.

```python
# ----------------------
# B5 a. Ambil transaksi dengan rating_pelanggan < 3
# ----------------------
df_rating_rendah = df[df['rating_pelanggan'] < 3]
print("=== B5a: Transaksi rating < 3 ===")
display(df_rating_rendah)
```

```python
# ----------------------
# B5 b. Jumlah & rata-rata total_bayar
# ----------------------
jumlah_transaksi = len(df_rating_rendah)
rata_rata_bayar = df_rating_rendah['total_bayar'].mean()

print("\n=== B5b ===")
print(f"Jumlah transaksi: {jumlah_transaksi}")
print(f"Rata-rata total_bayar: {rata_rata_bayar:,.2f}")
```

```python
# ----------------------
# B5 c. Kategori 'Kerajinan' ATAU 'Fashion' DAN jumlah >= 10
# ----------------------
df_kategori = df[
    ((df['kategori'] == 'Kerajinan') | (df['kategori'] == 'Fashion')) &
    (df['jumlah'] >= 10)
]
print("\n=== B5c: Kerajinan/Fashion & jumlah >= 10 ===")
display(df_kategori)
```

```python
# B6
hasil = df[
    (df['metode_bayar'] == 'Tunai') &
    (df['total_bayar'] > 500000)
]

hasil[['kota', 'produk', 'jumlah', 'total_bayar']] \
    .sort_values('total_bayar', ascending=False) \
    .head(10)
```

## Bagian C — Kolom turunan dan pengurutan

```python
df['harga_bruto'] = df['jumlah'] * df['harga_satuan']
df['hemat'] = df['harga_bruto'] - df['total_bayar']
df[['harga_bruto', 'total_bayar', 'hemat']].head()
```

```python
df['tanggal'] = pd.to_datetime(df['tanggal'])
df['bulan'] = df['tanggal'].dt.month
df['nama_hari'] = df['tanggal'].dt.day_name()
df[['tanggal', 'bulan', 'nama_hari']].head()
```

```python
# C3a
import numpy as np

df['kelas_nilai'] = np.where(
    df['total_bayar'] > 500000,
    'Besar',
    'Kecil'
)

# C3b
df['kelas_nilai'].value_counts()
```

```python
# C3c
df[['kota', 'produk', 'total_bayar', 'harga_bruto', 'hemat']] \
    .sort_values('hemat', ascending=False) \
    .head(5)
```

## Bagian D — Agregasi dengan groupby
Pola yang akan Anda pakai sepanjang semester: **split → apply → combine**.

```python
# DEMO D1 — agregasi sederhana
df.groupby('kategori')['total_bayar'].mean().round(0)
```

```python
# DEMO D2 — beberapa agregasi sekaligus
ringkas = (df.groupby('kategori')
             .agg(jumlah_transaksi=('id_transaksi','count'),
                  total_penjualan=('total_bayar','sum'),
                  rata_rata_nilai=('total_bayar','mean'),
                  rata_rata_rating=('rating_pelanggan','mean'))
             .round(2)
             .sort_values('total_penjualan', ascending=False))
ringkas
```

```python
# DEMO D3 — dua kunci pengelompokan
df.groupby(['kategori','metode_bayar'])['total_bayar'].sum().unstack(fill_value=0)
```

```python
# D4a
df.groupby('kota')['total_bayar'] \
    .sum() \
    .sort_values(ascending=False)
```

```python
# D4b
rating_kategori = df.groupby('kategori')['rating_pelanggan'] \
    .mean() \
    .sort_values()

rating_kategori
```

```python
rating_kategori.head(1)
```

```python
# D4c
penjualan_bulan = df.groupby('bulan')['total_bayar'] \
    .sum() \
    .sort_values(ascending=False)

penjualan_bulan
```
